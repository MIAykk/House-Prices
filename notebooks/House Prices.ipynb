## Data Cleaning


```python
df_clean=df.copy()
```

```python
missing_values=df_clean.isnull().sum().sort_values(ascending=False)
missing_values=missing_values[missing_values>0]
missing_values
```

```python
missing_percent=(df_clean.isnull().sum()/len(df_clean))*100
missing_percent
missing_df=pd.DataFrame({"Missing Count":df.isnull().sum(),"Missing Percent":missing_percent})
missing_df=missing_df[missing_df["Missing Count"]>0]
missing_df=missing_df.sort_values("Missing Percent",ascending=False)
missing_df

```

### Obeservation
Several features contain missing values.Columns such as PoolQC ,MiscFeature,Alley,and Fence have a very high percentage of missing data,while others such as LotFrontage have a moderate amount of mssing values.Appropriate strategies will be applied base on the (type) and proportion of missinf values.

## LotFrontage

```python
df_clean.groupby("Neighborhood")["LotFrontage"].median()
```

```python
df_clean["LotFrontage"]=(df_clean.groupby("Neighborhood")["LotFrontage"].transform(lambda x :x.fillna(x.median())))
```

### Observation
The "LotFrontage" column contains approximately 17% missing values.Since lot frontage varies across neighborhood,imputing missing values with the overall median could introduce bias.Therefore,missing values were filled using the median "LotFrontage" within each "Neighborhood".

## BsmtFinType2

```python
df_clean[df_clean["BsmtFinType2"].isnull()][["BsmtQual","BsmtCond","BsmtExposure","BsmtFinType1","BsmtFinType2","TotalBsmtSF"]]
```

```python
df_clean["BsmtFinType2"]=df_clean["BsmtFinType2"].fillna("None")
```

```python
df_clean["BsmtFinType2"].isnull().sum()
```

### Observation
Most missing values in BsmtFinType2 correspond to houses without a basement and were filled with "None" . One remaining missing value belonged to a house with an existing basement (TotalBsmtSF>2) , indicating a true missing value,which was imputed using the mode of the column
