## LotFrontage

```python
df_clean.groupby("Neighborhood")["LotFrontage"].median()
```

```python
df_clean["LotFrontage"]=(df_clean.groupby("Neighborhood")["LotFrontage"].transform(lambda x :x.fillna(x.median())))
```

### Observation
The "LotFrontage" column contains approximately 17% missing values.Since lot frontage varies across neighborhood,imputing missing values with the overall median could introduce bias.Therefore,missing values were filled using the median "LotFrontage" within each "Neighborhood".
